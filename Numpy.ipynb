# Why Use NumPy?

In Python we have lists that serve the purpose of arrays, but they are slow to process.

NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.

The array object in NumPy is called ndarray, it provides a lot of supporting functions that make working with ndarray very easy.

Arrays are very frequently used in data science, where speed and resources are very important

## Indexing on 2D array(matrices)

Format/Syntax: arr[row,column]or arr[row][column]

```python
arr5=np.random.randint(10,100,12)
arr5
```

```python
new_arr=arr5.reshape(3,4)
new_arr
```

```python
new_arr[-1,3]
```

```python
new_arr[2,1]
```

```python
new_arr[0,2]
```

```python
new_arr[0:2,1:3]
```

```python
new_arr[:,[2,0]]  #from all rows fetch index 2 and 0 columns
```

```python
new_arr[1]  #returns row with index 1
```

```python
ans=new_arr[:,3] #returns column with index 3
ans
```

```python
ans_re=ans.reshape((3,1))
ans_re
```

```python
new_arr[1:,1:]
```

```python
new_arr[:2,1:]
```

## Fancy Indexing

Selecting entire row or column in any order

```python
arr6=np.arange(1,21)
arr6
```

```python
reshape_arr=arr6.reshape(5,4)
reshape_arr
```

```python
reshape_arr[[3,1,4]]  #fetching rows of index 3, 1 & 4 -Retrieve 3rd row1st,1st row 2nd and 4th row 3rd
```

```python
reshape_arr[:,[3,0,1]] #fetching columns of index 3, 0 & 1
```
